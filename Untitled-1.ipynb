```python
import tensorflow as tf
import pickle
from tensorflow.keras.preprocessing.sequence import pad_sequences
import re
from bs4 import BeautifulSoup
import emoji
import numpy as np

# بارگذاری tokenizer
with open("tokenizer.pkl", "rb") as f:
    tokenizer = pickle.load(f)

def clean_text(text):
    if not isinstance(text, str):
        return ""
    
    text = text.lower()
    text = re.sub(r'http\S+|www\S+', '', text)  # حذف URLs
    text = BeautifulSoup(text, 'html.parser').get_text()  # حذف تگ‌های html
    text = emoji.demojize(text)  # تبدیل ایموجی‌ها به کلمه
    text = re.sub(r'[^\w\s]', '', text)  # حذف punctuation
    text = re.sub(r'\s+', ' ', text).strip()  # حذف فاصله‌های اضافی
    
    return text  # ✅ مهم: مقدار را برگردانید

def preprocess_text(texts, tokenizer, max_len=99):

    if isinstance(texts, str):
        texts = [texts]
    
    text_seq = tokenizer.texts_to_sequences(texts)
    text_padded = pad_sequences(text_seq, maxlen=max_len, padding="post")
    
    return text_padded

def Predict(text, model, tokenizer, max_len=99):
    # پاکسازی متن (فقط یک بار)
    cleaned_text = clean_text(text)
    
    # تبدیل به لیست برای preprocess
    text_padded = preprocess_text(cleaned_text, tokenizer, max_len)
    
    # پیش‌بینی
    y_prob = model.predict(text_padded, verbose=0)
    y_pred = np.argmax(y_prob, axis=1)
    
    classes = ['منفی😒', 'مثبت😃', 'طبیعی و خنثی🙂']
    pred_class = classes[y_pred[0]]
    pred_prob = y_prob[0][y_pred[0]]
    
    return pred_class, pred_prob

# بارگذاری مدل
loaded_model = tf.keras.models.load_model('LSTM_Sentiment_analysis.h5')

```

```python
# تست کردن
new_text = "I'm happy"
pred_class, prob = Predict(new_text, loaded_model, tokenizer, max_len=99)
print(f"Text: {new_text}")
print(f"Prediction: {pred_class} (confidence: {prob:.4f})")
```
